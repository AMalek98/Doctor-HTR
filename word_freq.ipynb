```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from collections import Counter
import os

def load_medical_terminology_data(file_path="enter your path here"):
    """
    Loads medical terminology data from a text file containing labeled medical terms.
    
    Reads a structured text file where each line contains labeled medical terminology,
    typically in the format "line_id: medical_term" or similar structured format.
    
    Args:
        file_path (str): Path to the text file containing medical terminology data
    
    Returns:
        list: Extracted medical terms from the dataset
    
    Raises:
        FileNotFoundError: If the specified file cannot be found
        ValueError: If the file format is invalid or cannot be processed
    """
    # Path to medical terminology data file
    file_path = "enter your path here"  # Path to medical terms text file
    
    print(f"📂 Loading medical terminology data from: {file_path}")
    
    try:
        if not os.path.exists(file_path):
            raise FileNotFoundError(f"Data file not found: {file_path}")
        
        # Read the file and process the lines
        with open(file_path, 'r', encoding='utf-8') as file:
            lines = file.readlines()
        
        print(f"📊 Read {len(lines)} lines from data file")
        
        # Extract medical terms (words after the colon)
        medical_terms = []
        processed_lines = 0
        
        for line in lines:
            line = line.strip()
            if ':' in line:
                # Extract the medical terminology after the colon
                medicine_name = line.split(':', 1)[1].strip()
                
                # Split compound medicine names into individual terms
                terms = medicine_name.split()
                medical_terms.extend(terms)
                processed_lines += 1
            elif line:  # Handle lines without colons but with content
                terms = line.split()
                medical_terms.extend(terms)
                processed_lines += 1
        
        print(f"✅ Successfully processed {processed_lines} lines")
        print(f"📋 Extracted {len(medical_terms)} medical terms")
        
        # Filter out very short terms (likely not meaningful medical terms)
        filtered_terms = [term for term in medical_terms if len(term) >= 2]
        
        if len(filtered_terms) != len(medical_terms):
            print(f"🔍 Filtered out {len(medical_terms) - len(filtered_terms)} terms shorter than 2 characters")
        
        return filtered_terms
        
    except FileNotFoundError as e:
        print(f"❌ File not found: {e}")
        print("💡 Please verify the file path and ensure the file exists")
        raise
        
    except Exception as e:
        print(f"❌ Error processing file: {e}")
        raise

def calculate_frequency_statistics(medical_terms):
    """
    Calculates comprehensive frequency statistics for medical terminology.
    
    Computes detailed frequency analysis including term counts, distribution
    statistics, and identifies patterns in medical vocabulary usage.
    
    Args:
        medical_terms (list): List of medical terms extracted from the dataset
    
    Returns:
        dict: Comprehensive frequency analysis results including DataFrame and statistics
    """
    if not medical_terms:
        raise ValueError("No medical terms provided for analysis")
    
    print(f"📊 Calculating frequency statistics for {len(medical_terms)} terms...")
    
    # Count the frequency of each medical term
    term_counts = Counter(medical_terms)
    
    # Create comprehensive DataFrame
    frequency_df = pd.DataFrame(term_counts.items(), columns=['medicine', 'frequency'])
    frequency_df = frequency_df.sort_values('frequency', ascending=False).reset_index(drop=True)
    
    # Calculate statistical metrics
    frequency_stats = {
        'total_terms': len(medical_terms),
        'unique_terms': len(term_counts),
        'max_frequency': frequency_df['frequency'].max(),
        'min_frequency': frequency_df['frequency'].min(),
        'mean_frequency': frequency_df['frequency'].mean(),
        'median_frequency': frequency_df['frequency'].median(),
        'std_frequency': frequency_df['frequency'].std()
    }
    
    # Calculate distribution insights
    single_occurrence = sum(1 for count in term_counts.values() if count == 1)
    high_frequency = sum(1 for count in term_counts.values() if count >= 10)
    
    frequency_stats.update({
        'terms_single_occurrence': single_occurrence,
        'terms_high_frequency': high_frequency,
        'single_occurrence_percentage': (single_occurrence / len(term_counts)) * 100,
        'high_frequency_percentage': (high_frequency / len(term_counts)) * 100
    })
    
    print("\\n📈 Frequency Analysis Results:")
    print(f"   Total Terms: {frequency_stats['total_terms']}")
    print(f"   Unique Terms: {frequency_stats['unique_terms']}")
    print(f"   Most Frequent Term: '{frequency_df.iloc[0]['medicine']}' ({frequency_stats['max_frequency']} occurrences)")
    print(f"   Mean Frequency: {frequency_stats['mean_frequency']:.2f}")
    print(f"   Terms with Single Occurrence: {frequency_stats['terms_single_occurrence']} ({frequency_stats['single_occurrence_percentage']:.1f}%)")
    print(f"   High Frequency Terms (≥10): {frequency_stats['terms_high_frequency']} ({frequency_stats['high_frequency_percentage']:.1f}%)")
    
    return {
        'dataframe': frequency_df,
        'statistics': frequency_stats,
        'counter': term_counts
    }

def display_frequency_table(frequency_analysis, top_n=10):
    """
    Displays formatted frequency table with top medical terms.
    
    Args:
        frequency_analysis (dict): Results from calculate_frequency_statistics
        top_n (int): Number of top terms to display
    """
    frequency_df = frequency_analysis['dataframe']
    
    print(f"\\n📋 Top {top_n} Most Frequent Medical Terms:")
    print("=" * 50)
    
    for i, row in frequency_df.head(top_n).iterrows():
        print(f"{i+1:2d}. {row['medicine']:<20} : {row['frequency']:3d} occurrences")
    
    print("\\n📋 Complete Frequency Distribution:")
    print(frequency_df.to_string(index=False))

# Execute medical terminology analysis
print("🏥 Medical Terminology Frequency Analysis")
print("=" * 50)

try:
    # Load medical terminology data
    medical_terms = load_medical_terminology_data()
    
    if medical_terms:
        # Calculate frequency statistics
        frequency_analysis = calculate_frequency_statistics(medical_terms)
        
        # Display results
        display_frequency_table(frequency_analysis, top_n=10)
        
        print("\\n✅ Medical terminology analysis completed successfully")
        
    else:
        print("⚠️ No medical terms found in the data file")
        
except Exception as e:
    print(f"❌ Analysis failed: {e}")
    frequency_analysis = None
```

```python
def create_frequency_visualizations(frequency_analysis, save_plots=False, output_dir="enter your path here"):
    """
    Creates comprehensive visualizations of medical terminology frequency distribution.
    
    Generates multiple chart types to visualize the distribution patterns of
    medical terms, including top frequent terms and least frequent terms analysis.
    
    Args:
        frequency_analysis (dict): Results from calculate_frequency_statistics
        save_plots (bool): Whether to save plots to files
        output_dir (str): Directory to save plot files
    
    Returns:
        None
    """
    if frequency_analysis is None:
        print("❌ Cannot create visualizations - no frequency analysis data available")
        return
    
    frequency_df = frequency_analysis['dataframe']
    stats = frequency_analysis['statistics']
    
    print("\\n📊 Creating Medical Terminology Frequency Visualizations...")
    print("-" * 60)
    
    # Configuration for plots
    plt.style.use('default')
    
    # 1. Top 10 Most Frequent Medical Terms
    print("📈 Generating Top 10 Most Frequent Terms Chart...")
    
    top_10 = frequency_df.head(10)
    
    plt.figure(figsize=(14, 8))
    bars = plt.bar(range(len(top_10)), top_10['frequency'], color='skyblue', alpha=0.8, edgecolor='navy')
    
    # Customize the plot
    plt.xlabel('Medical Terms', fontsize=12, fontweight='bold')
    plt.ylabel('Frequency (Number of Occurrences)', fontsize=12, fontweight='bold')
    plt.title('Top 10 Most Frequent Medical Terms\\nDataset Vocabulary Analysis', fontsize=14, fontweight='bold', pad=20)
    
    # Set x-axis labels with rotation for readability
    plt.xticks(range(len(top_10)), top_10['medicine'], rotation=45, ha='right')
    
    # Add frequency values on top of bars
    for i, bar in enumerate(bars):
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 0.5,
                f'{int(height)}', ha='center', va='bottom', fontweight='bold')
    
    # Add grid for better readability
    plt.grid(axis='y', alpha=0.3, linestyle='--')
    
    # Add statistics text box
    stats_text = f"""Dataset Statistics:
    Total Terms: {stats['total_terms']:,}
    Unique Terms: {stats['unique_terms']:,}
    Mean Frequency: {stats['mean_frequency']:.1f}
    Max Frequency: {stats['max_frequency']}"""
    
    plt.text(0.02, 0.98, stats_text, transform=plt.gca().transAxes, 
             fontsize=9, verticalalignment='top', 
             bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
    
    plt.tight_layout()
    
    if save_plots:
        output_path = "enter your path here"  # Path to save top frequent terms plot
        try:
            plt.savefig(output_path, dpi=300, bbox_inches='tight')
            print(f"💾 Top frequent terms plot saved to: {output_path}")
        except Exception as e:
            print(f"⚠️ Failed to save plot: {e}")
    
    plt.show()
    
    # 2. Least Frequent Medical Terms (Bottom 10)
    print("\\n📉 Generating Least Frequent Terms Chart...")
    
    # Get terms with frequency = 1 (if available) or bottom 10
    bottom_terms = frequency_df[frequency_df['frequency'] == 1].head(10)
    if len(bottom_terms) < 5:
        bottom_terms = frequency_df.tail(10)
    
    plt.figure(figsize=(14, 8))
    bars = plt.bar(range(len(bottom_terms)), bottom_terms['frequency'], color='orange', alpha=0.8, edgecolor='darkred')
    
    # Customize the plot
    plt.xlabel('Medical Terms', fontsize=12, fontweight='bold')
    plt.ylabel('Frequency (Number of Occurrences)', fontsize=12, fontweight='bold')
    plt.title('Least Frequent Medical Terms\\nRare Vocabulary Analysis', fontsize=14, fontweight='bold', pad=20)
    
    # Set x-axis labels with rotation for readability
    plt.xticks(range(len(bottom_terms)), bottom_terms['medicine'], rotation=45, ha='right')
    
    # Add frequency values on top of bars
    for i, bar in enumerate(bars):
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 0.02,
                f'{int(height)}', ha='center', va='bottom', fontweight='bold')
    
    # Add grid for better readability
    plt.grid(axis='y', alpha=0.3, linestyle='--')
    
    # Add insights text box
    single_occurrence_count = stats['terms_single_occurrence']
    insights_text = f"""Rare Terms Analysis:
    Single Occurrence: {single_occurrence_count} terms ({stats['single_occurrence_percentage']:.1f}%)
    Min Frequency: {stats['min_frequency']}
    These terms may need special attention in training"""
    
    plt.text(0.02, 0.98, insights_text, transform=plt.gca().transAxes, 
             fontsize=9, verticalalignment='top', 
             bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.8))
    
    plt.tight_layout()
    
    if save_plots:
        output_path = "enter your path here"  # Path to save least frequent terms plot
        try:
            plt.savefig(output_path, dpi=300, bbox_inches='tight')
            print(f"💾 Least frequent terms plot saved to: {output_path}")
        except Exception as e:
            print(f"⚠️ Failed to save plot: {e}")
    
    plt.show()
    
    # 3. Frequency Distribution Histogram
    print("\\n📊 Generating Frequency Distribution Histogram...")
    
    plt.figure(figsize=(12, 8))
    
    # Create histogram of frequencies
    frequencies = frequency_df['frequency'].values
    bins = min(30, len(np.unique(frequencies)))  # Adaptive bin count
    
    plt.hist(frequencies, bins=bins, color='lightgreen', alpha=0.7, edgecolor='darkgreen')
    
    plt.xlabel('Frequency (Number of Occurrences)', fontsize=12, fontweight='bold')
    plt.ylabel('Number of Terms', fontsize=12, fontweight='bold')
    plt.title('Medical Terms Frequency Distribution\\nDataset Vocabulary Spread Analysis', fontsize=14, fontweight='bold', pad=20)
    
    # Add vertical lines for mean and median
    plt.axvline(stats['mean_frequency'], color='red', linestyle='--', linewidth=2, label=f'Mean: {stats["mean_frequency"]:.1f}')
    plt.axvline(stats['median_frequency'], color='blue', linestyle='-.', linewidth=2, label=f'Median: {stats["median_frequency"]:.1f}')
    
    plt.legend()
    plt.grid(alpha=0.3)
    
    # Add distribution insights
    distribution_text = f"""Distribution Insights:
    Standard Deviation: {stats['std_frequency']:.2f}
    Range: {stats['min_frequency']} - {stats['max_frequency']}
    High Frequency Terms (≥10): {stats['terms_high_frequency']} ({stats['high_frequency_percentage']:.1f}%)"""
    
    plt.text(0.98, 0.98, distribution_text, transform=plt.gca().transAxes, 
             fontsize=9, verticalalignment='top', horizontalalignment='right',
             bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.8))
    
    plt.tight_layout()
    
    if save_plots:
        output_path = "enter your path here"  # Path to save distribution histogram
        try:
            plt.savefig(output_path, dpi=300, bbox_inches='tight')
            print(f"💾 Distribution histogram saved to: {output_path}")
        except Exception as e:
            print(f"⚠️ Failed to save plot: {e}")
    
    plt.show()

def generate_comprehensive_report(frequency_analysis):
    """
    Generates a comprehensive analysis report with insights and recommendations.
    
    Args:
        frequency_analysis (dict): Results from frequency analysis
    
    Returns:
        dict: Comprehensive report with insights and recommendations
    """
    if frequency_analysis is None:
        return {'status': 'failed', 'reason': 'no_analysis_data'}
    
    frequency_df = frequency_analysis['dataframe']
    stats = frequency_analysis['statistics']
    
    # Generate insights
    insights = {
        'dataset_overview': {
            'total_medical_terms': stats['total_terms'],
            'unique_vocabulary_size': stats['unique_terms'],
            'vocabulary_diversity_ratio': stats['unique_terms'] / stats['total_terms'],
            'most_common_term': frequency_df.iloc[0]['medicine'],
            'highest_frequency': stats['max_frequency']
        },
        'distribution_analysis': {
            'mean_frequency': stats['mean_frequency'],
            'median_frequency': stats['median_frequency'],
            'frequency_standard_deviation': stats['std_frequency'],
            'skewness_indicator': 'High' if stats['mean_frequency'] > stats['median_frequency'] * 1.5 else 'Moderate'
        },
        'vocabulary_patterns': {
            'high_frequency_terms': stats['terms_high_frequency'],
            'single_occurrence_terms': stats['terms_single_occurrence'],
            'coverage_imbalance': stats['single_occurrence_percentage'] > 50,
            'training_challenges': 'High vocabulary diversity with many rare terms'
        }
    }
    
    # Generate recommendations
    recommendations = {
        'model_training': [
            'Consider class weighting for balanced learning of rare vs frequent terms',
            'Implement curriculum learning starting with frequent terms',
            'Use data augmentation for terms with single occurrences',
            'Consider subword tokenization for rare medical terms'
        ],
        'dataset_optimization': [
            'Collect more samples for rare medical terms if possible',
            'Validate accuracy of single-occurrence terms',
            'Consider domain-specific preprocessing for medical abbreviations',
            'Implement stratified sampling for training/validation splits'
        ],
        'evaluation_strategy': [
            'Test model performance separately on frequent vs rare terms',
            'Monitor recall specifically for medical terms with clinical importance',
            'Implement medical domain-specific evaluation metrics',
            'Validate against medical terminology databases'
        ]
    }
    
    # Print comprehensive report
    print("\\n" + "="*80)
    print("📋 COMPREHENSIVE MEDICAL TERMINOLOGY ANALYSIS REPORT")
    print("="*80)
    
    print("\\n📊 Dataset Overview:")
    print(f"   Total Medical Terms: {insights['dataset_overview']['total_medical_terms']:,}")
    print(f"   Unique Vocabulary: {insights['dataset_overview']['unique_vocabulary_size']:,}")
    print(f"   Vocabulary Diversity: {insights['dataset_overview']['vocabulary_diversity_ratio']:.3f}")
    print(f"   Most Common Term: '{insights['dataset_overview']['most_common_term']}' ({insights['dataset_overview']['highest_frequency']} occurrences)")
    
    print("\\n📈 Distribution Analysis:")
    print(f"   Mean Frequency: {insights['distribution_analysis']['mean_frequency']:.2f}")
    print(f"   Median Frequency: {insights['distribution_analysis']['median_frequency']:.2f}")
    print(f"   Standard Deviation: {insights['distribution_analysis']['frequency_standard_deviation']:.2f}")
    print(f"   Distribution Skewness: {insights['distribution_analysis']['skewness_indicator']}")
    
    print("\\n🎯 Vocabulary Patterns:")
    print(f"   High Frequency Terms (≥10): {insights['vocabulary_patterns']['high_frequency_terms']}")
    print(f"   Single Occurrence Terms: {insights['vocabulary_patterns']['single_occurrence_terms']}")
    print(f"   Coverage Imbalance: {'Yes' if insights['vocabulary_patterns']['coverage_imbalance'] else 'No'}")
    
    print("\\n🚀 Recommendations:")
    print("\\n   Model Training:")
    for rec in recommendations['model_training']:
        print(f"     • {rec}")
    
    print("\\n   Dataset Optimization:")
    for rec in recommendations['dataset_optimization']:
        print(f"     • {rec}")
    
    print("\\n   Evaluation Strategy:")
    for rec in recommendations['evaluation_strategy']:
        print(f"     • {rec}")
    
    print("="*80)
    
    return {
        'insights': insights,
        'recommendations': recommendations,
        'status': 'completed'
    }

# Execute visualization and reporting if analysis was successful
if frequency_analysis is not None:
    print("\\n🎨 Generating Visualizations and Comprehensive Report...")
    
    # Create frequency visualizations
    create_frequency_visualizations(frequency_analysis, save_plots=False)
    
    # Generate comprehensive report
    final_report = generate_comprehensive_report(frequency_analysis)
    
    print("\\n✅ Medical Terminology Frequency Analysis Pipeline Completed!")
    print("📊 All visualizations and analysis reports have been generated")
    
else:
    print("⚠️ Cannot generate visualizations and report - frequency analysis incomplete")
```
